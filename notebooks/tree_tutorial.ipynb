# Create a tree
To create a tree using this DraftTree class, you need to instantiate it by passing a single argument: a sequence (like a list or tuple) of integers representing the parent of each node.
Based on the __init__ method in your code, the list you pass in must follow a strict set of rules to satisfy the breadth-first (non-decreasing depth) requirement.
The Rules for the parents List
1. The Root Node is First: The very first element in your list (index 0) must be -1. This represents the root of the tree.
2. Parents Precede Children: For any subsequent node at index u, its parent's value must be a valid previous index (i.e., 0 <= parent < u).
3. Breadth-First Order: You must list the nodes layer by layer. You cannot list a node of depth 2, and then follow it with a new node of depth 1.

---

### Examples
Here is how you would map different tree structures into the required list format.

#### Example 1: A Simple Chain (Linear Tree)
If you want a straightforward chain (e.g., Root -> Node 1 -> Node 2 -> Node 3), where each node is the parent of the next:
```python
# Node index:    0   1  2  3
# Parent index: -1   0  1  2
parents_list = [-1,  0, 1, 2]

tree = DraftTree(parents_list)
```

#### Example 2: A Star Graph (Flat Tree)
If you want the root node to have three direct children, and no other branches:
```python
# Node index:    0   1  2  3
# Parent index: -1   0  0  0
parents_list = [-1,  0, 0, 0]

tree = DraftTree(parents_list)
```

#### Example 3: A Balanced Binary Tree
If you want a tree where the root has two children, and those children each have two children:
```
       0 (Root)
      / \
     1   2
    / \ / \
   3  4 5  6
```

To build this, you map out the parents for each index in breadth-first order (layer by layer):
- 0: Root (-1)
- 1: Child of 0
- 2: Child of 0
- 3: Child of 1
- 4: Child of 1
- 5: Child of 2
- 6: Child of 2

```python
parents_list = [-1, 0, 0, 1, 1, 2, 2]
tree = DraftTree(parents_list)
```

What happens if you break the order?
If you try to declare a tree where the depths bounce around—for example, [-1, 0, 1, 0] (which attempts to place a depth-1 child after a depth-2 child)—the class will calculate the depths as [0, 1, 2, 1] and immediately raise this ValueError:
ValueError: nodes must be listed in breadth-first (non-decreasing depth) order
